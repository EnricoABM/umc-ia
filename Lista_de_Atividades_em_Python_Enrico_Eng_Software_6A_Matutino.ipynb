# **7. Ordenação de Lista**
  * Enunciado: Desenvolva um programa que receba uma lista de números do usuário e ordene essa lista em ordem crescente. O programa deve exibir a lista antes e depois da ordenação.

```python
# Menu
print("==================")
print("ORDENAÇÃO DE LISTA")
print("==================")

# Entrada do usuário
print("Digite uma lista de números separador por um espaço")
lista = input("Lista: ").split()

try:
  # Validação de tipo de dado
  for l in lista:
    if not l.isdigit():
      raise ValueError("Valor inválido!!!")

  # Conversão de tipo
  lista = list(map(lambda x: float(x), lista))

  # Ordenação da lista
  lista_ord = sorted(lista)

  # Apresentação
  print(f"Lista: {lista}")
  print(f"Lista Ordenada: {lista_ord}")

except ValueError as e:
  # Encerrando em caso de tipo de dado inválido
  print(e)
  print("Encerrando...")
```

# **8. Verificador de Números Primos**
Enunciado: Crie um programa que determine se um número fornecido pelo usuário é primo. O programa deve informar se o número é primo ou não e listar todos os divisores do número.

```python
# Menu
print("=============================")
print("VERIFICADOR DE NÚMEROS PRIMOS")
print("=============================")

# Entrada do usuário
numero = input("Digite um número: ")

# Validar entrada
while not numero.isdigit():
  numero = input("Valor inválido.\nDigite um número: ")

# Conversão de tipos
numero = int(numero)

# Lista de divisores
divisores = []

# Quantidade de divisores
qntd_divs = 0

# Percorre todos os números anteriores a numero
for n in range(1, numero+1):
  # Verifica se possui divisores
  if numero % n == 0:
    qntd_divs += 1
    divisores.append(n)

# Se possui 2 divisores é primo, caso contrário, não.
print(f"{numero} É PRIMO" if qntd_divs <= 2 else f"{numero} NÃO É PRIMO")
print("\nDivisores: ")
for div in divisores:
  print(div, end=" ")
```

# **9. Sistema de Login Simples**
* Enunciado: Implemente um sistema de login simples que armazene os nomes de usuário e senhas em um dicionário. O programa deve permitir que o usuário se cadastre e faça login, validando as credenciais inseridas.

```python
class App:
  # Inicia a lista de usuários
  def __init__(self):
    self.users = {}

  # Apresenta a tela inicial
  def mainMenuShow(self):
    while True:
      # Menu
      print("========================")
      print("SISTEMA DE LOGIN SIMPLES")
      print("========================")
      print("[1] - Cadastrar Usuário")
      print("[2] - Login")
      print("[3] - Sair")

      # Entrada do usuário
      op = input("Informe um valor: ")

      # Validar entrada
      while op not in ("1", "2", "3"):
        op = input("Valor inválido.\n Informe um valor: ")

      # Opções
      if (op == "1"):
        self.createUserMenu()
      elif (op == "2"):
        self.loginMenu()
      elif (op == "3"):
        break


  def createUserMenu(self):
    # Menu
    print("========================")
    print("SISTEMA DE LOGIN SIMPLES")
    print("========================")
    print("CADASTRAR USUÁRIO")

    # Entradas do usuários
    name = input("Informe o nome do usuário: ")
    password = input("Informe a senha do usuário: ")

    # Salvando usuário no Dict
    self.save(name, password)
    print("Cadastrado com Sucesso!!!")

  def save(self, name, password):
    # Salva o usuário no Dict
    self.users[name] = password

  def loginMenu(self):
    # Menu
    print("========================")
    print("SISTEMA DE LOGIN SIMPLES")
    print("========================")
    print("LOGIN")

    # Entrada do usuário
    name = input("Informe o nome do usuário: ")
    password = input("Informe a senha do usuário: ")

    # Valida as credenciais
    login = self.validateUser(name, password)

    # Verifica se as credenciais foram validas
    if login:
      self.userMenu(name)
    else:
      print("Login Incorreto")


  def validateUser(self, name, password):
    # Verifica se existe o usuário no dicionário
    stored_password = self.users[name]

    # Caso não exista, retorna false
    if (stored_password is None):
      return False

    # Verifica a senha, caso o usuário exista
    if stored_password == password:
      return True
    else:
      return False

  def userMenu(self, name):
    # Menu de Login
    while True:
      print(f"Olá {name}")
      input("Aperte Enter para sair...")
      break


# Execução do App
app = App()
app.mainMenuShow()
```

# **10. Calculadora de Fatorial**
* Enunciado: Escreva um programa que calcule o fatorial de um número fornecido pelo usuário utilizando um loop for. O programa deve exibir o resultado do fatorial ao final.

```python
# Menu
print("=======================")
print("CALCULADORA DE FATORIAL")
print("=======================")

# Entrada
numero = input("Informe um Número: ")

# Validação da entrada
while not numero.isdigit():
  print("Valor Inválido!")
  numero = input("Informe um Número: ")

# Conversão
numero = int(numero)

# Inicia a variável
resultado = 1

# Cria a sequência de números do fatorial
numeros = range(1, numero + 1)

# Percorre a sequência e multiplica pelo atual
for n in numeros[::-1]:
  resultado *= n

# Apresentação do resultado
print(f"Fatorial de {numero} = {resultado}")
```

# **14. Calculadora de Média Ponderada**
* Enunciado: Implemente uma calculadora que permita ao usuário calcular a média ponderada de um conjunto de notas. O usuário deve inserir as notas e os respectivos pesos, e o programa deve calcular e exibir a média ponderada.

```python
# Importação da biblioteca
import math

# Valida a entrada
def validar_numero(msg: str):
  inp = input(msg)
  while not inp.isdigit():
    print("Valor inválido!!!")
    inp = input(msg)

  return float(inp)

# Menu
print("==============================")
print("CALCULADORA DE MÉDIA PONDERADA")
print("==============================")

# Inicia as variáveis
denominadores = 0
numeradores = 0
count = 0

while True:
  # Entrada de valores
  print(f"Quantidade de valores na média {count}")
  valor = validar_numero("Informe o valor: ")
  peso = validar_numero("Informe o peso: ")

  # Incrementa as variáveis
  numeradores += valor * peso
  denominadores += peso

  # Continuidade
  op = input("Deseja continuar? [S/N]")
  if (op.upper() == "N"):
    break

  count += 1

# Apresenta o resultado
print(f"\nMédia: {numeradores/denominadores:.2f}")
```

* Enunciado: Crie um programa que simule um caixa eletrônico. O usuário deve inserir o valor que deseja sacar, e o programa deve exibir quantas notas de cada valor serão entregues (R$100, R$50, R$20, R$10).

```python
# Validação da entrada
def validar_numero(msg: str):
  inp = input(msg)
  while not inp.isdigit():
    print("Valor inválido!!!")
    inp = input(msg)

  return float(inp)

# Menu
print("=============================")
print("SIMULADOR DE CAIXA ELETRÔNICO")
print("=============================")

# Entrada de usuários
valor = validar_numero("Valor a sacar: R$")

# Dicionários de notas
notas = {
    100: 0,
    50: 0,
    20: 0,
    10: 0,
    5: 0,
    2: 0
  }
# Variável auxiliar
resto = valor

# Itera sobre cada nota
for nota in notas:
  # Pega a divisão inteira e incrementa a quantidade de notas
  notas[nota] += resto // nota
  # Atualiza a variável com o resto da divisão
  resto = resto % nota

# Apresenta as notas
print("\nNOTAS")
for nota, qtnd in notas.items():
  print(f"R{nota}: {qtnd:.0f}", end=" / ")

# Aprenseta o restante
print("\nResto: " + str(resto))
```
